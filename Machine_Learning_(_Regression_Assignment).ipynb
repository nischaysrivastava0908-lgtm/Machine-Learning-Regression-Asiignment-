# Machine Learning ( Regression Assignment )

1. What is Simple Linear Regression?

Answer :
Simple Linear Regression is a statistical method used to understand the relationship between one independent variable (X) and one dependent variable (Y). It predicts the value of Y using a straight-line equation, \(Y=mX+c\), where m represents the slope and c represents the intercept.

2. What are the key assumptions of Simple Linear Regression?

Answer :
The key assumptions are linearity, independence of errors, constant variance of errors (homoscedasticity), normally distributed residuals for statistical inference, and no significant outliers that distort the model. These assumptions help ensure that the estimated coefficients, predictions, and statistical conclusions are reliable.

3. What does the coefficient m represent in the equation Y = mX + c?

Answer :
The coefficient m represents the slope of the regression line. It indicates how much the predicted dependent variable (Y) changes when the independent variable (X) increases by one unit. A positive slope indicates an increasing relationship, while a negative slope indicates a decreasing relationship.

4. What does the intercept c represent in the equation Y = mX + c?

Answer :
The intercept c represents the predicted value of Y when X equals zero. It indicates where the regression line crosses the Y-axis. Its practical meaning depends on the context and whether X = 0 is a meaningful value within the observed data range.

5. How do we calculate the slope m in Simple Linear Regression?

Answer :
The slope is calculated by dividing the covariance between X and Y by the variance of X. The formula is \(m=\frac{\sum(X_i-\bar X)(Y_i-\bar Y)}{\sum(X_i-\bar X)^2}\). It measures the average change in the predicted Y for every one-unit increase in X.

6. What is the purpose of the least squares method in Simple Linear Regression?

Answer :
The least squares method finds the best-fitting regression line by minimizing the sum of squared differences between actual and predicted values. These differences are called residuals. Squaring them prevents positive and negative errors from canceling out and penalizes larger prediction errors more heavily.

7. How is the coefficient of determination (R²) interpreted in Simple Linear Regression?

Answer :
R² measures the proportion of variation in the dependent variable explained by the regression model. Its value typically ranges from 0 to 1 for a model with an intercept. For example, an R² of 0.80 means the model explains 80% of the observed variation in Y.

8. What is Multiple Linear Regression?

Answer :
Multiple Linear Regression is a statistical technique that predicts a dependent variable using two or more independent variables. It estimates the relationship between the dependent variable and each predictor while holding the other included predictors constant. Its general equation is \(Y=b_0+b_1X_1+b_2X_2+\cdots+b_nX_n+\epsilon\).

9. What is the main difference between Simple and Multiple Linear Regression?

Answer :
Simple Linear Regression uses only one independent variable to predict a dependent variable, whereas Multiple Linear Regression uses two or more independent variables. Multiple regression can account for several factors simultaneously, providing a more comprehensive model when multiple predictors contribute to the outcome.

10. What are the key assumptions of Multiple Linear Regression?

Answer :
The key assumptions include a linear relationship between predictors and the expected outcome, independent errors, constant error variance, no perfect multicollinearity, and normally distributed errors for conventional statistical inference. Correct model specification and the absence of highly influential observations are also important for reliable interpretation.

11. What is heteroscedasticity, and how does it affect Multiple Linear Regression?

Answer :
Heteroscedasticity occurs when the variance of regression errors is not constant across observations. It can make conventional standard errors, confidence intervals, and hypothesis tests unreliable. Ordinary least squares coefficients may remain unbiased under suitable exogeneity conditions, but they may be inefficient. Robust standard errors or weighted least squares can help address it.

12. How can you improve a Multiple Linear Regression model with high multicollinearity?

Answer :
High multicollinearity occurs when independent variables are strongly correlated. It can make coefficient estimates unstable and increase standard errors. It can be addressed by removing redundant predictors, combining related variables, collecting more data, or using Ridge or Lasso regression. Variance Inflation Factor (VIF) can help identify problematic predictors.

13. What are some common techniques for transforming categorical variables for use in regression models?

Answer :
Categorical variables can be converted into numerical representations using techniques such as one-hot encoding, dummy encoding, ordinal encoding, and effect coding. One-hot encoding creates separate binary columns for categories, while ordinal encoding assigns ordered numbers. The appropriate method depends on whether the categories have a meaningful order.

14. What is the role of interaction terms in Multiple Linear Regression?

Answer :
Interaction terms allow the effect of one independent variable on the dependent variable to depend on another independent variable. For example, an interaction between advertising expenditure and season may indicate that advertising effectiveness changes across seasons. Interaction terms help capture relationships that cannot be represented by additive effects alone.

15. How can the interpretation of intercept differ between Simple and Multiple Linear Regression?

Answer :
In Simple Linear Regression, the intercept represents the predicted outcome when the single predictor equals zero. In Multiple Linear Regression, it represents the predicted outcome when all predictors equal zero. In both cases, the intercept may have limited practical meaning if those zero values are unrealistic or outside the observed data range.

16. What is the significance of the slope in regression analysis, and how does it affect predictions?

Answer :
The slope represents the expected change in the dependent variable for a one-unit increase in a predictor, holding other predictors constant in multiple regression. A positive slope increases predicted outcomes, while a negative slope decreases them. Its magnitude determines how strongly changes in the predictor affect the model's predictions.

17. How does the intercept in a regression model provide context for the relationship between variables?

Answer :
The intercept establishes the model's baseline prediction when all independent variables equal zero. It provides a reference point for understanding how predictors influence the outcome relative to that baseline. However, if zero is unrealistic for the predictors, centering variables around their means can make the intercept more meaningful.

18. What are the limitations of using R² as a sole measure of model performance?

Answer :
R² measures how much variation the model explains, but it does not establish causation, guarantee accurate predictions on new data, or confirm that regression assumptions are satisfied. It can increase when additional predictors are added, even when they contribute little value. Adjusted R², residual analysis, and test-set metrics provide additional insight.

19. How would you interpret a large standard error for a regression coefficient?

Answer :
A large standard error indicates that the estimated coefficient has considerable uncertainty relative to its scale. It may result from multicollinearity, limited data, high residual variability, or weak information about the predictor's effect. A large standard error generally produces wider confidence intervals and makes it harder to establish statistical significance.

20. How can heteroscedasticity be identified in residual plots, and why is it important to address it?

Answer :
Heteroscedasticity can be identified by plotting residuals against fitted values or predictors. A funnel-shaped pattern or systematically increasing or decreasing residual spread suggests nonconstant variance. Addressing it matters because conventional standard errors and hypothesis tests may be misleading. Robust standard errors or weighted least squares are common solutions.

21. What does it mean if a Multiple Linear Regression model has a high R² but low adjusted R²?

Answer :
A high R² with a much lower adjusted R² may indicate that the model contains too many predictors that contribute little explanatory value. R² never decreases when predictors are added to an ordinary least squares model with an intercept, whereas adjusted R² penalizes unnecessary complexity. This difference suggests reviewing the predictors and model specification.

22. Why is it important to scale variables in Multiple Linear Regression?

Answer :
Scaling transforms variables to comparable numerical ranges, often through standardization or normalization. It can improve numerical stability and make coefficient magnitudes easier to compare when variables use different units. Scaling is especially useful for regularized regression methods such as Ridge and Lasso, where the penalty depends on coefficient size.

23. What is Polynomial Regression?

Answer :
Polynomial Regression is a regression technique that models a nonlinear relationship between an independent variable and a dependent variable by including powers of the predictor. It fits a polynomial curve rather than restricting predictions to a straight line. For example, a quadratic model includes both X and \(X^2\) as predictors.

24. How does Polynomial Regression differ from Linear Regression?

Answer :
Simple Linear Regression models a straight-line relationship between X and Y, while Polynomial Regression can model curved relationships using higher powers of X. Although polynomial regression is nonlinear in the predictor, it remains linear in its coefficients. Higher-degree polynomials provide flexibility but can increase the risk of overfitting.

25. When is Polynomial Regression used?

Answer :
Polynomial Regression is used when the relationship between a predictor and the outcome follows a curved pattern that a straight line cannot adequately represent. Examples include modeling growth trends, changes in demand, and physical processes. It is useful when residual plots suggest curvature and the additional complexity improves validation performance.

26. What is the general equation for Polynomial Regression?

Answer :
The general polynomial regression equation of degree n is \(Y=b_0+b_1X+b_2X^2+\cdots+b_nX^n+\epsilon\). Here, \(b_0\) is the intercept, \(b_1,\ldots,b_n\) are the coefficients, X is the independent variable, and \(\epsilon\) represents the error term. The degree determines the maximum power of X included.

27. Can Polynomial Regression be applied to multiple variables?

Answer :
Yes, Polynomial Regression can be extended to multiple variables by including powers and interaction terms of different predictors. For example, a model may contain \(X_1^2\), \(X_2^2\), and \(X_1X_2\). This allows it to capture curvature and interactions, but the number of features can grow rapidly as predictors and polynomial degree increase.

28. What are the limitations of Polynomial Regression?

Answer :
Polynomial Regression can overfit training data, especially at high degrees, and may produce unrealistic predictions outside the observed range. Higher powers can also cause numerical instability and make interpretation difficult. Choosing an appropriate degree, scaling predictors, using regularization, and evaluating performance on validation data can reduce these problems.

29. What methods can be used to evaluate model fit when selecting the degree of a polynomial?

Answer :
Common methods include cross-validation, adjusted R², RMSE, MAE, and residual analysis. Cross-validation evaluates how well different polynomial degrees generalize to unseen data, while adjusted R² accounts for model complexity. The best degree generally balances predictive accuracy and simplicity rather than merely maximizing training R².


30. Why is visualization important in Polynomial Regression?

Answer :
Visualization helps reveal whether the relationship between predictors and the outcome is curved and whether a polynomial model fits that pattern. Plotting observed data alongside fitted curves can expose underfitting, overfitting, outliers, and unrealistic behavior near the boundaries. Residual plots also help determine whether important patterns remain unexplained.

31. How is Polynomial Regression implemented in Python?

Answer :
Polynomial Regression can be implemented in Python using PolynomialFeatures and LinearRegression from scikit-learn. First, transform the original predictor into polynomial features, then fit a linear regression model using those features. Finally, evaluate predictions on test data using metrics such as RMSE or R² to assess generalization.
